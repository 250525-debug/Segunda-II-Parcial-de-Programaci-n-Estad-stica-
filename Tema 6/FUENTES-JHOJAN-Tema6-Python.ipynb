```python
import numpy as np
import matplotlib.pyplot as plt

# Completa: construye el degradado de 5x5
fila = np.linspace(0, 255, 5)
gris = np.tile(fila, (5, 1))

# Completa: imprime shape, dtype y mean()
print("Shape:", gris.shape)
print("Dtype:", gris.dtype)
print("Mean:", gris.mean())
```

```python
# Completa: visualiza y guarda la imagen
# Convertimos 'gris' a una matriz 2D de 5x5 si es un vector 1D
gris_2d = np.tile(gris, (5, 1)) if gris.ndim == 1 else gris

plt.imshow(gris_2d, cmap="gray")
plt.savefig("gris_tema6.png")
plt.show()
```

```python
# Completa: recorte de la mitad derecha
recorte = gris[:, 2:]
print(recorte)
```

```python
# Completa: construye la imagen RGB de dos mitades
rgb = np.zeros((4,4,3), dtype=np.uint8)
rgb[:, 0:2] = [255, 0, 0]   # mitad izquierda: elige un color (R,G,B)
rgb[:, 2:4] = [0, 0, 255]   # mitad derecha: elige otro color
   
print(rgb.shape)
print("Pixel [0,0]:", rgb[0, 0])
print("Pixel [0,3]:", rgb[0, 3])
```

```python
# Completa: extrae cada canal
import matplotlib.pyplot as plt

# Completa: extrae cada canal
canal_r = rgb[:, :, 0]
canal_g = rgb[:, :, 1]
canal_b = rgb[:, :, 2]

# Mostrar los canales (opcional, pero recomendado para verificar)
plt.figure(figsize=(10, 3))

plt.subplot(1, 3, 1)
plt.imshow(canal_r, cmap='Reds')
plt.title('Canal R (Rojo)')

plt.subplot(1, 3, 2)
plt.imshow(canal_g, cmap='Greens')
plt.title('Canal G (Verde)')

plt.subplot(1, 3, 3)
plt.imshow(canal_b, cmap='Blues')
plt.title('Canal B (Azul)')

plt.show()
```

```python
# Completa: gris_promedio y gris_luminosidad
gris_promedio = (canal_r + canal_g + canal_b) / 3
gris_luminosidad = 0.299 * canal_r + 0.587 * canal_g + 0.114 * canal_b

print("--- Gris Promedio ---")
print(gris_promedio)
print("\n--- Gris Luminosidad ---")
print(gris_luminosidad)
```

```python
# Completa: binarizacion con tu propio umbral
umbral = 50 
binaria = gris_luminosidad > umbral

binaria = np.where(gris_luminosidad > umbral, 255, 0).astype(np.uint8)

print("Imagen binarizada:")
print(binaria)

plt.imshow(binaria, cmap="gray", vmin=0, vmax=255)
plt.title("Imagen binarizada")
plt.axis("off")
plt.show()
```
