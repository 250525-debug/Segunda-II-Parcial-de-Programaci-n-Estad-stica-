```python
from PIL import Image, ImageDraw

# Crear una imagen de 120 x 120 píxeles con fondo blanco
im = Image.new("RGB", (120, 120), color=(255, 255, 255))
draw = ImageDraw.Draw(im)

# 1. Rectángulo amarillo
draw.rectangle((10, 10, 50, 40), fill=(255, 255, 0))

# 2. Cuadrado celeste
draw.rectangle((70, 10, 100, 40), fill=(135, 206, 235))

# 3. Círculo verde
draw.ellipse((10, 60, 40, 90), fill=(0, 128, 0))

# 4. Línea roja
draw.line((65, 65, 105, 100), fill=(255, 0, 0), width=5)

# Guardar imagen
im.save("mi_imagen.png")

# Mostrar resultados
display(im)
print("Imagen creada correctamente.")
print("Tamaño:", im.size)
print("Formato:", im.format)
print("Modo:", im.mode)
```

```python
# Completa: las cuatro transformaciones
from PIL import Image

# Abrir imagen original
im = Image.open("mi_imagen.png")

# 1. Convertir a escala de grises
img_gris = im.convert("L")
img_gris.save("mi_imagen_gris.png")

# 2. Rotar 90 grados
img_rotada = im.rotate(90, expand=True)
img_rotada.save("mi_imagen_rotada.png")

# 3. Reducir a la mitad
img_reducida = im.resize((im.width // 2, im.height // 2))
img_reducida.save("mi_imagen_reducida.png")

# 4. Voltear verticalmente
img_volteada = im.transpose(Image.Transpose.FLIP_TOP_BOTTOM)
img_volteada.save("mi_imagen_volteada.png")

# Mostrar tamaños
print("Tamaño original:", im.size)
print("Tamaño gris:", img_gris.size)
print("Tamaño rotado:", img_rotada.size)
print("Tamaño reducido:", img_reducida.size)
print("Tamaño volteado:", img_volteada.size)

# Mostrar transformaciones
display(img_gris)
display(img_rotada)
display(img_reducida)
display(img_volteada)
```

```python
from PIL import Image, ImageEnhance
  # Completa: dos factores de brillo distintos
   
# Abrir imagen original
im = Image.open("mi_imagen.png").convert("RGB")

# Ajustar brillo
brillo_bajo = ImageEnhance.Brightness(im).enhance(0.5)
brillo_alto = ImageEnhance.Brightness(im).enhance(1.5)

# Ajustar contraste
contraste_bajo = ImageEnhance.Contrast(im).enhance(0.5)
contraste_alto = ImageEnhance.Contrast(im).enhance(1.5)

# Guardar imágenes
brillo_bajo.save("brillo_bajo.png")
brillo_alto.save("brillo_alto.png")
contraste_bajo.save("contraste_bajo.png")
contraste_alto.save("contraste_alto.png")

# Mostrar resultados
display(brillo_bajo)
display(brillo_alto)
display(contraste_bajo)
display(contraste_alto)

print("Imágenes de brillo y contraste guardadas.")
```

```python
from PIL import Image, ImageFilter
# Completa: aplica y guarda ambos filtros
# Abrir imagen original
im = Image.open("mi_imagen.png").convert("RGB")

# Filtro gaussiano
gaussiana = im.filter(ImageFilter.GaussianBlur(radius=2))
gaussiana.save("mi_imagen_gaussiana.png")

# Filtro de detección de bordes
bordes = im.filter(ImageFilter.FIND_EDGES)
bordes.save("mi_imagen_bordes.png")

# Mostrar resultados
display(gaussiana)
display(bordes)

print("Filtro gaussiano y detección de bordes completados.")
```

```python
import numpy as np
from PIL import Image
from skimage import color
  # Completa: array, rgb2gray, comparacion con convert('L')

# Abrir imagen original
im = Image.open("mi_imagen.png").convert("RGB")

# Convertir imagen a arreglo NumPy
array_rgb = np.array(im)

# Conversión con scikit-image
gris_skimage = color.rgb2gray(array_rgb)

# Conversión con Pillow
gris_pillow = np.array(im.convert("L"))

# Comparar el mismo píxel
x, y = 20, 20

valor_skimage = gris_skimage[y, x] * 255
valor_pillow = gris_pillow[y, x]

print("Píxel analizado:", (x, y))
print("Valor scikit-image:", valor_skimage)
print("Valor Pillow:", valor_pillow)
print("Diferencia absoluta:",
      abs(valor_skimage - valor_pillow))
```

```python
import numpy as np
from PIL import Image
from skimage import color, filters
# Completa: threshold_otsu y binarizacion
# Abrir imagen original
im = Image.open("mi_imagen.png").convert("RGB")

# Convertir a arreglo y escala de grises
array_rgb = np.array(im)
gris = color.rgb2gray(array_rgb)

# Calcular umbral de Otsu
threshold_otsu = filters.threshold_otsu(gris)

# Binarizar imagen
imagen_binaria = gris > threshold_otsu

# Calcular proporción de píxeles True
proporcion_true = np.mean(imagen_binaria)

# Guardar imagen binaria
imagen_guardar = (imagen_binaria * 255).astype(np.uint8)
Image.fromarray(imagen_guardar).save("mi_imagen_binaria.png")

# Imprimir resultados
print("Umbral de Otsu:", threshold_otsu)
print("Proporción de píxeles True:", proporcion_true)
print("Porcentaje de píxeles True:", proporcion_true * 100, "%")

# Mostrar imagen binaria
display(Image.open("mi_imagen_binaria.png"))
```
