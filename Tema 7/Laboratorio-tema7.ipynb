```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = 'jaguar.jpg'

# Verificar si la imagen existe en el entorno, si no, crear una imagen sintética de prueba
if os.path.exists(image_path):
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
else:
    print(f"Advertencia: No se encontró '{image_path}'. Generando una imagen de prueba temporal.")
    # Crear una imagen RGB de ejemplo (degradado)
    yy, xx = np.mgrid[0:300, 0:300]
    r = (xx / 300.0) * 255
    g = (yy / 300.0) * 255
    b = ((300 - xx) / 300.0) * 255
    image_rgb = np.stack([r, g, b], axis=-1).astype(np.uint8)

scale_factor_1 = 3.0
scale_factor_2 = 1/3.0
height, width = image_rgb.shape[:2]

new_height = int(height * scale_factor_1)
new_width = int(width * scale_factor_1)
zoomed_image = cv2.resize(src=image_rgb,
                          dsize=(new_width, new_height),
                          interpolation=cv2.INTER_CUBIC)

new_height1 = int(height * scale_factor_2)
new_width1 = int(width * scale_factor_2)
scaled_image = cv2.resize(src=image_rgb,
                          dsize=(new_width1, new_height1),
                          interpolation=cv2.INTER_AREA)

fig, axs = plt.subplots(1, 3, figsize=(10, 4))
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image\nShape:'+str(image_rgb.shape))
axs[1].imshow(zoomed_image)
axs[1].set_title('Zoomed Image\nShape:'+str(zoomed_image.shape))
axs[2].imshow(scaled_image)
axs[2].set_title('Scaled Image\nShape:'+str(scaled_image.shape))

for ax in axs:
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import os

image_path = 'jaguar.jpg'

# Validar si existe la imagen para evitar errores de lectura
if os.path.exists(image_path):
    img = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
else:
    print(f"No se encontró la imagen en {image_path}. Usando una imagen negra de prueba.")
    image_rgb = np.zeros((300, 300, 3), dtype=np.uint8)

# Calcular matriz de rotación
center = (image_rgb.shape[1] // 2, image_rgb.shape[0] // 2)
angle = 30
scale = 1
rotation_matrix = cv2.getRotationMatrix2D(center, angle, scale)
rotated_image = cv2.warpAffine(image_rgb, rotation_matrix, (image_rgb.shape[1], image_rgb.shape[0]))

# Graficar resultados
fig, axs = plt.subplots(1, 2, figsize=(7, 4))
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image')
axs[1].imshow(rotated_image)
axs[1].set_title('Image Rotation')

for ax in axs:
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os

image_path = 'jaguar.jpg'

# Verificar si la imagen existe
if os.path.exists(image_path):
    img = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
else:
    print(f"No se encontró {image_path}. Usando imagen de prueba temporal.")
    image_rgb = np.zeros((300, 300, 3), dtype=np.uint8)

width, height = image_rgb.shape[1], image_rgb.shape[0]
tx, ty = 100, 70

translation_matrix = np.array([[1, 0, tx], [0, 1, ty]], dtype=np.float32)
translated_image = cv2.warpAffine(image_rgb, translation_matrix, (width, height))

fig, axs = plt.subplots(1, 2, figsize=(7, 4))
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image')

axs[1].imshow(translated_image)
axs[1].set_title('Image Translation')

for ax in axs:
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = 'jaguar.jpg'

if os.path.exists(image_path):
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
else:
    image_rgb = np.zeros((300, 300, 3), dtype=np.uint8)

b, g, r = cv2.split(image_rgb)
b_normalized = cv2.normalize(b.astype('float'), None, 0, 1, cv2.NORM_MINMAX)
g_normalized = cv2.normalize(g.astype('float'), None, 0, 1, cv2.NORM_MINMAX)
r_normalized = cv2.normalize(r.astype('float'), None, 0, 1, cv2.NORM_MINMAX)

normalized_image = cv2.merge((b_normalized, g_normalized, r_normalized))
print(normalized_image[:, :, 0])

plt.imshow(normalized_image)
plt.xticks([])
plt.yticks([])
plt.title('Normalized Image')
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = 'jaguar.jpg'

if os.path.exists(image_path):
    img = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
else:
    image_rgb = np.zeros((300, 300, 3), dtype=np.uint8)

edges = cv2.Canny(image_rgb, 100, 200)

fig, axs = plt.subplots(1, 2, figsize=(7, 4))
axs[0].imshow(image_rgb)
axs[0].set_title('Original Image')
axs[1].imshow(edges, cmap='gray')
axs[1].set_title('Image Edges')

for ax in axs:
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

image_path = 'jaguar.jpg'

if os.path.exists(image_path):
    image = cv2.imread(image_path)
    image_gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
else:
    image_gray = np.zeros((300, 300), dtype=np.uint8)

kernel = np.ones((3, 3), np.uint8)
dilated = cv2.dilate(image_gray, kernel, iterations=2)
eroded = cv2.erode(image_gray, kernel, iterations=2)
opening = cv2.morphologyEx(image_gray, cv2.MORPH_OPEN, kernel)
closing = cv2.morphologyEx(image_gray, cv2.MORPH_CLOSE, kernel)

fig, axs = plt.subplots(2, 2, figsize=(7, 7))
axs[0, 0].imshow(dilated, cmap='gray')
axs[0, 0].set_title('Dilated Image')
axs[0, 1].imshow(eroded, cmap='gray')
axs[0, 1].set_title('Eroded Image')
axs[1, 0].imshow(opening, cmap='gray')
axs[1, 0].set_title('Opening')
axs[1, 1].imshow(closing, cmap='gray')
axs[1, 1].set_title('Closing')

for ax in axs.flatten():
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
plt.show()
```
