```python
from PIL import Image, ImageDraw, ImageEnhance, ImageFilter
import numpy as np
img = Image.new("RGB", (100, 100), color=(30, 30, 120))
draw = ImageDraw.Draw(img)
draw.rectangle([10, 10, 45, 45], fill=(255, 165, 0)) # cuadrado naranja
draw.ellipse([55, 55, 90, 90], fill=(0, 200, 100)) # circulo verde
print(img.size, img.mode)
img.save("tema7_demo.png")
```

```python
img2 = Image.open("tema7_demo.png")
print(img2.format, img2.size, img2.mode)
print("Pixel de fondo (5,5):", img2.getpixel((5, 5)))
print("Pixel del cuadrado (20,20):", img2.getpixel((20, 20)))
print("Pixel del circulo (70,70):", img2.getpixel((70, 70)))
```

```python
# Python — celda 3
img_gris = img2.convert("L")
img_rot = img2.rotate(45, expand=True)
img_pequena = img2.resize((50, 50))
img_volteada = img2.transpose(Image.FLIP_LEFT_RIGHT)
recorte = img2.crop((0, 0, 50, 50))
print("Gris - mode:", img_gris.mode)
print("Rotada 45 - size:", img_rot.size)
print("Redimensionada - size:", img_pequena.size)
print("Volteada - size:", img_volteada.size)
print("Recorte - size:", recorte.size)
```

```python
print("Original, pixel donde estaba el cuadrado (20,20):", img2.getpixel((20, 20)))
print("Volteada, mismo punto (20,20):", img_volteada.getpixel((20, 20)))
print("Volteada, punto simetrico (79,20):", img_volteada.getpixel((79, 20)))
```

```python
realzador = ImageEnhance.Brightness(img2)
img_brillante = realzador.enhance(1.5)
img_super = realzador.enhance(3.0)
print("Fondo original:", img2.getpixel((5,5)), "-> x1.5:", img_brillante.getpixel((5,5)))
print("Cuadrado original:", img2.getpixel((20,20)), "-> x3.0:", img_super.getpixel((20,20)))
```

```python
img_blur = img2.filter(ImageFilter.GaussianBlur(radius=3))
img_blur.save("tema7_blur.png")
img_bordes = img2.filter(ImageFilter.FIND_EDGES)
img_bordes.save("tema7_bordes.png")
```

```python
from skimage import color, filters
arr = np.array(img2)
print(arr.shape, arr.dtype)
gris_sk = color.rgb2gray(arr)
gris_pil = np.array(img2.convert("L"))
print("Pillow .convert('L'), fondo (5,5):", gris_pil[5, 5])
print("skimage rgb2gray, fondo (5,5):", round(gris_sk[5, 5], 4), "x255 =", round(gris_sk[5,5]*255, 2))
```

```python
suavizada = filters.gaussian(arr, sigma=2, channel_axis=-1)
bordes = filters.sobel(gris_sk)
umbral_otsu = filters.threshold_otsu(gris_sk)
binaria = gris_sk > umbral_otsu
print("Umbral de Otsu calculado automaticamente:", round(umbral_otsu, 4))
print("Tipo de binaria:", binaria.dtype)
print("Proporcion de pixeles 'True' (por encima del umbral):", round(binaria.mean(), 4))
```
